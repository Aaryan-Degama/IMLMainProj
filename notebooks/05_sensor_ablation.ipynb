# 5. Sensor ablation

LR compares both sensors (58 features), back only (29) and thigh only (29). MLP fixed-split experiments are excluded.

LR is the reference model for sensor ablation because it achieved the highest mean LOSO accuracy and macro F1 close to XGBoost. RF and XGBoost ablations are supported but excluded from the reported ablation.


```python
from pathlib import Path
import os
import sys
ROOT = next((p for p in [Path.cwd(), Path.cwd().parent]
             if (p / 'run_sensor_ablation.py').exists()), None)
if ROOT is None:
    raise FileNotFoundError('Project root not found.')
os.environ.setdefault('OPENBLAS_NUM_THREADS', '1')
os.environ.setdefault('OMP_NUM_THREADS', '1')
sys.path.insert(0, str(ROOT))
from IPython.display import display, Image
from run_sensor_ablation import run_ablation, OUT
```

## Controls

Cached windows, labels, LOSO folds, settings and seeds are identical. Combined-sensor results are reused; ablation outputs are separate.


```python
MODEL_NAMES = ('LR',)
DURATIONS = (1, 2)
SENSORS = ('back', 'thigh')
summary, differences = run_ablation(MODEL_NAMES, DURATIONS, SENSORS)
display(summary.round(4))
display(differences.round(4))
```

## Sensor performance

Scores average subjects equally. Positive single-minus-combined differences favor the single sensor. Subject-bootstrap intervals are exploratory.


```python
for filename in ['sensor_macro_f1.png', 'sensor_accuracy.png', 'sensor_paired_differences.png']:
    display(Image(filename=str(OUT / 'figures' / filename)))
```

## Activity recall

Recall pools held-out windows. Recorded fitting times exclude feature extraction.


```python
for model in MODEL_NAMES:
    for seconds in DURATIONS:
        display(Image(filename=str(OUT / 'figures' / f'{model}_{seconds}s_class_recall.png')))
```
